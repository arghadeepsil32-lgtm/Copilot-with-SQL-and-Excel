Question 4 : Identify and handle missing values in the dataset using Copilot.

Answer ▶

```python
import pandas as pd

df = pd.read_csv("Excel_dataset.csv")

# Check missing values
print(df.isnull().sum())
```

```python
# For categorical columns, replace missing values with the mode (most frequent value).

df["Customer_Name"].fillna(df["Customer_Name"].mode()[0], inplace=True)
df["Region"].fillna(df["Region"].mode()[0], inplace=True)
df["Product"].fillna(df["Product"].mode()[0], inplace=True)

# For the numeric column Sales, replace missing values with the median.

df["Sales"].fillna(df["Sales"].median(), inplace=True)

# For Quantity, first convert invalid values to numeric and then fill missing values.

df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
df["Quantity"].fillna(df["Quantity"].median(), inplace=True)

# Step 3: Verify Missing Values Are Removed

print(df.isnull().sum())
```

Question 6 : Generate:

Column chart OR bar chart and Interpret which region has highest sales

Answer ▶

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Excel_dataset.csv")

# Total sales by region
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

print(region_sales)

region_sales.plot(
    kind="bar",
    figsize=(8,5),
    title="Total Sales by Region"
)

plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

Question 10 : Generate VBA code using Copilot for cleaning data (removing duplicates, filling null values,
format table).

Answer ▶


```python
Sub CleanData()

    Dim ws As Worksheet
    Dim LastRow As Long
    Dim LastCol As Long
    Dim DataRange As Range

    Set ws = ActiveSheet

    ' Find last row and column
    LastRow = ws.Cells(ws.Rows.Count, 1).End(xlUp).Row
    LastCol = ws.Cells(1, ws.Columns.Count).End(xlToLeft).Column

    Set DataRange = ws.Range(ws.Cells(1, 1), ws.Cells(LastRow, LastCol))

    ' Remove duplicate rows
    DataRange.RemoveDuplicates Columns:=Array(1, 2, 3, 4, 5, 6), Header:=xlYes

    ' Refresh last row after removing duplicates
    LastRow = ws.Cells(ws.Rows.Count, 1).End(xlUp).Row

    ' Fill blank cells with "N/A"
    On Error Resume Next
    DataRange.SpecialCells(xlCellTypeBlanks).Value = "N/A"
    On Error GoTo 0

    ' Format header row
    With ws.Range(ws.Cells(1, 1), ws.Cells(1, LastCol))
        .Font.Bold = True
        .Interior.Color = RGB(217, 225, 242)
    End With

    ' Apply borders
    DataRange.Borders.LineStyle = xlContinuous

    ' Auto-fit columns
    DataRange.Columns.AutoFit

    MsgBox "Data Cleaning Completed Successfully!", vbInformation

End Sub
```
